# Đánh giá mô hình (Evaluation)

```python
import ast
import re

# Chuẩn hóa cột label
def fix_label_format(label_str):
    fixed = re.sub(r'(?<=\[|\s|,)(O|B-[A-Z]+|I-[A-Z]+)(?=,|\s|\])', r'"\1"', label_str)
    return ast.literal_eval(fixed)

df_ner['label_fixed'] = df_ner['label'].apply(fix_label_format)
```

```python
df_ner['token_fixed'] = df_ner['text'].apply(lambda x: x.split())
```

```python
from transformers import pipeline, AutoTokenizer, AutoModelForTokenClassification

tokenizer = AutoTokenizer.from_pretrained("dslim/bert-base-NER")
model = AutoModelForTokenClassification.from_pretrained("dslim/bert-base-NER")
ner_pipeline = pipeline("ner", model=model, tokenizer=tokenizer, aggregation_strategy="simple", device=-1)  # dùng CPU

y_true = df_ner['label_fixed'].tolist()
y_pred = []

for tokens in df_ner['token_fixed']:
    text = " ".join(tokens)
    result = ner_pipeline(text)

    pred_labels = ['O'] * len(tokens)
    for entity in result:
        word = entity['word']
        label = entity['entity_group']

        for i, token in enumerate(tokens):
            if word.lower() in token.lower():
                pred_labels[i] = 'B-' + label
                break

    y_pred.append(pred_labels)
```

```python
from seqeval.metrics import classification_report

print(classification_report(y_true, y_pred))
```

```python
import re
import ast

# Chuẩn hóa nhãn và token
def fix_label_format(label_str):
    fixed = re.sub(r'(?<=\[|\s|,)(O|B-[A-Z]+|I-[A-Z]+)(?=,|\s|\])', r'"\1"', label_str)
    return ast.literal_eval(fixed)

def fix_token_format(token_str):
    token_str = token_str.replace("’", "'").replace("[", "").replace("]", "")
    raw_tokens = token_str.split(",")
    return [t.strip().strip('"').strip("'") for t in raw_tokens if t.strip()]

# Áp dụng vào DataFrame
df_finer['label_fixed'] = df_finer['label'].apply(fix_label_format)
df_finer['token_fixed'] = df_finer['token'].apply(fix_token_format)
```

```python
from transformers import pipeline, AutoTokenizer, AutoModelForTokenClassification

tokenizer = AutoTokenizer.from_pretrained("dslim/bert-base-NER")
model = AutoModelForTokenClassification.from_pretrained("dslim/bert-base-NER")
ner_pipeline = pipeline("ner", model=model, tokenizer=tokenizer, aggregation_strategy="simple", device=-1)
```

```python
y_true = df_finer['label_fixed'].tolist()
y_pred = []

for tokens in df_finer['token_fixed']:
    text = " ".join(tokens)
    result = ner_pipeline(text)

    pred_labels = []
    for token in tokens:
        label = 'O'
        for ent in result:
            if token.lower() in ent['word'].lower():
                label = 'B-' + ent['entity_group']
                break
        pred_labels.append(label)

    y_pred.append(pred_labels)
```

```python
y_true_aligned = []
y_pred_aligned = []

for yt, yp in zip(y_true, y_pred):
    if len(yt) == len(yp):
        y_true_aligned.append(yt)
        y_pred_aligned.append(yp)

from seqeval.metrics import classification_report
print(classification_report(y_true_aligned, y_pred_aligned))
```

#  Đánh giá kết quả NER trên hai tập dữ liệu FLARE

## 1. So sánh kết quả

| Loại dữ liệu     | Dataset `flare_ner` (ảnh 1) | Dataset `flare_finer_ord` (ảnh 2) |
|------------------|-----------------------------|------------------------------------|
| Tổng nhãn        | 891                         | 365                                |
| F1-score (micro) | **0.25**                    | **0.25**                           |
| F1-score (macro) | 0.21                        | **0.22**                           |
| Weighted F1      | 0.29                        | **0.29**                           |
| Nhãn mạnh nhất   | PER (precision: 0.89)       | ORG (recall: 0.50)                 |
| Nhãn yếu nhất    | MISC (không xuất hiện)      | MISC (không xuất hiện)             |

---

## 2. Nhận xét chi tiết

###  Dataset `flare_ner` (dạng câu):
- Mô hình cho precision **rất cao với nhãn PER** (`0.89`) nhưng recall thấp (`0.19`) → mô hình **chỉ đánh nhãn khi cực kỳ chắc chắn**.
- ORG và LOC có recall tương đối nhưng precision thấp → **khả năng phân biệt thực thể chưa ổn định**.
- Ngữ cảnh đầy đủ, nhưng dữ liệu chưa được token hóa rõ ràng nên khó đánh nhãn theo từ.

###  Dataset `flare_finer_ord` (dạng token):
- Precision toàn cục thấp hơn, nhưng recall cao hơn hẳn.
- **ORG (0.31 F1) và LOC (0.37 F1)** là hai nhãn được nhận diện tốt hơn.
- Dữ liệu rõ ràng hơn về định dạng token → phù hợp hơn với đánh giá hoặc fine-tuning mô hình.

---

## 3. Kết luận cuối cùng

- Mô hình `dslim/bert-base-NER` chỉ đạt khoảng **25% F1-score** → không đáp ứng yêu cầu cho trích xuất thực thể trong văn bản tài chính.
- Dữ liệu `flare_ner` có ngữ cảnh đầy đủ, thích hợp cho mô hình cần hiểu toàn câu.
- Dữ liệu `flare_finer_ord` gọn nhẹ, dễ huấn luyện, cho recall tốt hơn.
- Nhãn `MISC` bị bỏ qua hoàn toàn → cần làm giàu thêm nhãn này nếu huấn luyện mô hình mới.
